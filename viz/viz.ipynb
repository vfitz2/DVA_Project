```python
import pandas as pd
import polars as pl
from datetime import datetime, timedelta
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
feature_PATH = "../data/processed/features/feature_cols.csv"
feature_cols = pd.read_csv(feature_PATH)["feature"].tolist()

response_PATH = "../data/processed/response_cols/response_cols.csv"
response_cols = pd.read_csv(response_PATH)["response"].tolist()

train_data = pd.DataFrame() 
train_PATH = f"../data/processed/train/train_data.parquet"
train_data = pl.DataFrame(pd.concat([train_data, pd.read_parquet(train_PATH)]))

test_data = pd.DataFrame() 
test_PATH = f"../data/processed/test/test_data.parquet"
test_data = pl.DataFrame(pd.concat([test_data , pd.read_parquet(test_PATH)]))

results_PATH = "../results/raw/xgb_predictions_vs_actuals.csv"
results_data = pd.read_csv(results_PATH)
```

```python
train_results = pl.DataFrame(results_data[results_data["split"]=="train"])
test_results = pl.DataFrame(results_data[results_data["split"]=="test"])

train_results = pl.concat([train_data, train_results], how="horizontal")
test_results = pl.concat([test_data, test_results], how="horizontal")

df = test_results 
```

```python
display(df)
```

```python
df = (
    df
    .with_columns
    (
        pl.when(pl.col("stand")=="R").then(pl.col("plate_x"))
        .otherwise(-1*pl.col("plate_x"))
        .alias("plate_x_bat_norm")
    )
    .with_columns
    (
        (
            (pl.col("plate_z") - pl.col("sz_bot"))
            /(pl.col("sz_top") - pl.col("sz_bot"))
        ).alias("z_norm")
    )
)
```

```python
x_bins = [-0.83, -0.28, 0.28, 0.83]
z_bins = [0.0, 1/3, 2/3, 1.0]
```

```python
df = (
    df
    .with_columns
    (
        x_bin=pl.col("plate_x_bat_norm")
        .bin_intervals
        (
            intervals=x_bins,
            labels = False
        )
    
    )
    .with_columns
    (
        z_bin=pl.col("z_norm")
        .bin_intervals
        (
            intervals=z_bins,
            labels = False
        )
    
    )
        .with_columns
    (
        ((pl.col("x_bin") + pl.col("z_bin")*5))
        .alias("cell")
    
    )
    .drop("x_bin","z_bin")
)
```

```python
df = (
    df
        .with_columns
        (
            pl.col("actual").mean().over("cell")        
            .alias("wrong_rate_per_cell") 
        )
        .with_columns
        (
            pl.col("pred_label").mean().over("cell")        
            .alias("challenge_rate_per_cell") 
        )
        .with_columns
        (
            (
                ((pl.col("pred_label") == 1) & (pl.col("actual") == 1)).sum().over("cell")
            /  (pl.col("pred_label") == 1).sum().over("cell")
            )
            .alias("challenge_precision_per_cell") 
        )
        .with_columns
        (
            (
                pl.col("challenge_precision_per_cell")/pl.col("wrong_rate_per_cell")
            )
            .alias("challenge_lift_per_cell") 
        )
        .with_columns
        (
            (
                ((pl.col("pred_label") == 1) & (pl.col("actual") == 1)).sum().over("cell")
            )
            .alias("caught")
        )
        .with_columns
        (
            (
                ((pl.col("pred_label") == 1) & (pl.col("actual") == 1)).sum().over("cell")
                / (pl.col("actual") == 1).sum().over("cell")
            )
            .alias("recall")
        )
        .with_columns
        (
            (
                pl.col("cell").count().over("cell")
            )
            .alias("pitches_in_cell")
        )  
    # .select(["cell","pitches_in_cell","wrong_rate_per_cell","caught","recall", "challenge_precision_per_cell","challenge_lift_per_cell"])
    # .unique(subset="cell", keep="first")
    # .sort("cell")   
)
df
```

```python
viz_summary_df = (
    df
    .select(["cell","pitches_in_cell","wrong_rate_per_cell","challenge_rate_per_cell","caught","recall", "challenge_precision_per_cell","challenge_lift_per_cell"])
    .unique(subset="cell", keep="first")
    .sort("cell")    
)
viz_summary_df
```
